```python
%sql
create catalog if not exists development;
```

```python
%sql
create database if not exists development.data
```

```python
%sql
CREATE VOLUME if not EXISTS development.data.files;
```

```python
df=spark.read.format('csv').option('header','true').option('inferschema','true').load("/Volumes/development/data/files/orders/orders.csv")
display(df)
```

Loading Parquet file and JSON file

```python
df=spark.read.format('JSON').option('multiline','true').load("/Volumes/development/data/files/orders/order_json.json")
display(df)
```

```python
df=spark.read.format('Parquet').load("/Volumes/development/data/files/orders/orders.parquet")
display(df)
```

Selecting specific columns and Renaming

```python
display(df.select('customer_name','email'))
```

```python
from pyspark.sql.functions import *
display(df.select(col("customer_name").alias("Customer"),col('email').alias("Customer_Email")))
```

Adding New Column with existing one and Renaming the column

```python
from pyspark.sql.functions import *
display(df.withColumn("id",col("order_id")+10))
```

```python
display(df.withColumn("Country",lit('USA')))
```

```python
display(df.withColumnRenamed("customer_name","name"))
```
